```python
from gtts import gTTS
from IPython.display import Audio
```

```python
import pyttsx3
from IPython.display import Audio, display

# Text to convert
text = '''I'm a tech-savvy and detail-oriented individual with a background in data science, machine learning, and UI/UX design. I completed a 6-month UI/UX course on Coursera, where I designed wireframes and a high-fidelity prototype for a gaming e-commerce website, combining my passion for gaming with thoughtful design. I've worked on several machine learning projects, including customer churn prediction, credit card fraud detection, soil image analysis using GPT-4o, and various EDA and data visualization tasks.'''

# Initialize TTS engine
engine = pyttsx3.init()
engine.setProperty('rate', 150)
engine.setProperty('volume', 0.8)

# Select voice
voices = engine.getProperty('voices')
engine.setProperty('voice', voices[0].id)  # 0 = Male, 1 = Female (may vary by OS)

# Save to WAV (not MP3)
output_file = 'output_voice.wav'
engine.save_to_file(text, output_file)
engine.runAndWait()

# Play the audio in the notebook
display(Audio(output_file))
```
